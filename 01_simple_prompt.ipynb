```python
import ollama
```

```python
import os
from dotenv import load_dotenv, find_dotenv

env_path = find_dotenv()
if not env_path:
    raise FileNotFoundError("Could not find .env file. Please ensure it exists in the project directory.")

load_dotenv(env_path)
ollama_api_key = os.getenv("OLLAMA_API_KEY")
print(f'Ollama API Key: {ollama_api_key}')
```

```python
response = ollama.chat(
    model="llama3.2:latest",
    messages=[
        {"role": "user", "content": "Which is the capital city of France?"}
    ]
)
print(type(response))
print(response['message']['content'])
```
