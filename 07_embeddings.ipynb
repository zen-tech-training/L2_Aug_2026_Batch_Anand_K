```python
from langchain_ollama import OllamaEmbeddings

embeddings = OllamaEmbeddings(model="nomic-embed-text:latest")

input_text = "I like music"
words = input_text.split()
for word in words:
    word_embedding = embeddings.embed_query(word)
    print(f"Word: {word}, Embedding: {word_embedding}")
    print(f"\tlength: {len(word_embedding)}")
```

```python
# Example to save vectors into FAISS vector store
from langchain_community.vectorstores import FAISS
texts = ["I like music", "I like sports", "I like movies"]
vector_store = FAISS.from_texts(texts, embeddings)
print(f'Vector store using FAISS {vector_store}')
vector_store.save_local("faiss_index")
print('FAISS vector store is saved locally!!')
```

```python
# Example to save vectors into Chroma vector store
from langchain_community.vectorstores import Chroma
texts = ["I like music", "I like sports", "I like movies"]
vector_store = Chroma.from_texts(texts, embeddings)
print(f'Vector store using Chroma {vector_store}')

persist_directory = "chroma_db"

# Create a Chroma vector store from the text chunks and embeddings, and persist it to disk
vector_store = Chroma.from_texts(
    texts=texts,
    embedding=embeddings,
    persist_directory=persist_directory
)

print('Chroma vector store is saved locally!!')
```
