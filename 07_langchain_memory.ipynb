```python
from langchain_ollama import ChatOllama

model = ChatOllama(model="gpt-oss:120b-cloud")
print('LLM is ready')
```

```python
response_1 = model.invoke("I am Anand")
response_2 = model.invoke("What is my name?")

print(response_2)
```

```python
from langchain_classic.chains import ConversationChain
from langchain_classic.memory import ConversationBufferMemory

memory = ConversationBufferMemory()
conversation_chain = ConversationChain(llm=model, memory=memory)

response_1 = conversation_chain.invoke("I am Anand")
conversation_chain.invoke("I work with Zensar Technologies")
conversation_chain.invoke("When India got independence?")
conversation_chain.invoke("What is a square?")
response = conversation_chain.invoke("What is my name?")

print(response)
```

```python
from langchain_classic.chains import ConversationChain
from langchain_classic.memory import ConversationBufferWindowMemory

memory = ConversationBufferWindowMemory(k=10)
conversation_chain = ConversationChain(llm=model, memory=memory)

response_1 = conversation_chain.invoke("I am Anand")
conversation_chain.invoke("I work with Zensar Technologies")
conversation_chain.invoke("When India got independence?")
conversation_chain.invoke("What is a square?")
response = conversation_chain.invoke("What is my name?")

print(response['response'])
```

```python
# Example of ConversationSummaryMemory
from langchain_classic.chains import ConversationChain
from langchain_classic.memory import ConversationSummaryMemory

memory = ConversationSummaryMemory(llm=model, memory_key="history")
conversation_chain = ConversationChain(llm=model, memory=memory)

response_1 = conversation_chain.invoke("I am Anand")
conversation_chain.invoke("I work with Zensar Technologies")
conversation_chain.invoke("When India got independence?")
conversation_chain.invoke("What is a square?")
conversation_chain.invoke("What is my name?")

final_conversation_summary = memory.load_memory_variables({})['history']
print(f"Final conversation summary: {final_conversation_summary}")
```

```python
# Example of ConversationSummaryBufferMemory
from langchain_classic.chains import ConversationChain
from langchain_classic.memory import ConversationSummaryBufferMemory
memory = ConversationSummaryBufferMemory(llm=model, max_token_limit=1)
conversation_chain = ConversationChain(llm=model, memory=memory)

conversation_chain.predict(input="I am Anand")  
conversation_chain.predict(input="I work with Zensar Technologies")
conversation_chain.predict(input="When India got independence?")
conversation_chain.predict(input="What is a square?")
conversation_chain.predict(input="What is my name?")

#final_conversation_summary = memory.load_memory_variables({})['history']
#print(f"Final conversation summary: {final_conversation_summary}")

print(conversation_chain.memory.summary)
```
